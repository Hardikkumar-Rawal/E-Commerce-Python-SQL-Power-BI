```python
# Remove duplicate records

df.drop_duplicates(inplace=True)
```

```python
#6 Replace invalid values (Replace N/A, NULL, " " with NaN

df=df.replace(['N/A','NULL',' '],np.nan)
```

```python
df
```

```python
#7 Remove extra spaces

df = df.apply(lambda x: x.str.strip() if x.dtype == 'object' else x)
```

```python
df
```

```python
#8 Standardize text format (Proper case)

df["Customer_Name"]=df["Customer_Name"].str.title()

df["City"]=df["City"].str.title()

df["State"]=df["State"].str.title()
```

```python
#9 Validate Email addresses

df = df[df['Email'].str.contains('@',na=False)].copy()
```

```python
#10 Convert date columns

df['Order_Date'] = pd.to_datetime(df['Order_Date'], format = 'mixed', dayfirst= True, errors='coerce')
df['Delivery_Date'] = pd.to_datetime(df['Delivery_Date'], format = 'mixed', dayfirst= True, errors='coerce')
```

```python
#11 Convert columns into numeric-columns

cols = ['Qty','Unit_Price','Discount']

for c in cols:
    df[c]= pd.to_numeric(df[c],errors= 'coerce')
```

```python
#12 Remove invalid quantity

df= df[df['Qty']>0]
```

```python
#13 Fill missing values

df['Discount'] = df['Discount'].fillna(0)

df['Phone'] = df['Phone'].fillna('Unknown')

df['Delivery_Date'] = df['Delivery_Date'].fillna(df['Order_Date'])
```

```python
#14 Feature engineering

# Create new columns

df['Sales']= df['Qty'] * df['Unit_Price']
df['Net_Amount']= df['Sales'] - (df['Sales'] * df['Discount']/100)
df['Profit']= df['Net_Amount'] * 0.2
df['Month']= df['Order_Date'].dt.month_name()
df['Year']= df['Order_Date'].dt.year
df['Weekday']= df['Order_Date'].dt.day_name()
```
